# TextParser Demo

Shows flat-mode parsing of a plain-text file. Tree mode is not supported and raises `NotImplementedError`.

| Aspect | Notes |
|--------|-------|
| **Pros** | Zero dependencies; instant; works on any `.txt` file regardless of content; entire file is a single ContentBlock ready to chunk |
| **Cons** | No structure at all — headings, sections, and tables are treated as plain text; one large block may produce poor chunk boundaries |
| **vs. MarkdownParser** | If your text file has Markdown-style headings, switch to MarkdownParser to get proper heading-scoped tree structure |
| **vs. chunking directly** | Wrapping in Document/Chunk schemas gives you a consistent interface across all file types — same chunker works on .txt and .pdf without code changes |

## Imports

```python
# Standard Library
import pathlib

# Third Party Library

# Private Library
from cleave.parsers.factory import ParserFactory
from cleave.schemas import Document
```

## Fixture

```python
FIXTURES = pathlib.Path.cwd().parent.parent / "tests" / "fixtures"
TXT_PATH = str(FIXTURES / "sample.txt")

if not (FIXTURES / "sample.txt").exists():
    from tests.fixtures.txt import create_sample_txt
    create_sample_txt(FIXTURES / "sample.txt")

print("TXT:", TXT_PATH)
```

| Mode | `Document` field | Use when |
|------|-----------------|----------|
| `flat` | `.pages` — one virtual `DocumentPage` | Any plain-text file; single-block extraction |
| `tree` | — not supported — | N/A — use MarkdownParser for structured plain text |

## Flat mode

```python
flat_doc_parser = ParserFactory.create(TXT_PATH, mode="flat")
flat_doc = flat_doc_parser.parse()

assert isinstance(flat_doc, Document)
print(f"Source type : {flat_doc.source.type}")
print(f"Pages       : {len(flat_doc.pages)}")
print(f"Page number : {flat_doc.pages[0].page_number}")  # None
print(f"Blocks      : {len(flat_doc.pages[0].blocks)}")
```

```python
print("Full text:\n")
print(flat_doc.full_text)
```

```python
for block in flat_doc.pages[0].blocks:
    print(f"  type={block.type}  pos={block.position}  content={block.content[:60]!r}")
```

## Tree mode — NotImplementedError

```python
try:
    TextParser(TXT_PATH, mode="tree")
except NotImplementedError as e:
    print(f"NotImplementedError raised as expected:\n  {e}")
```

## Default mode is flat

```python
default_doc = TextParser(TXT_PATH).parse()
assert default_doc.pages is not None
print("Default mode produces a flat document — pages is not None.")
```
