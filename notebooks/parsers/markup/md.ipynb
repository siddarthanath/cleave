# MarkdownParser Demo

Shows flat and tree parsing of a Markdown file, including heading hierarchy, table extraction, and the bridge to a flat chunker.

| Aspect | Notes |
|--------|-------|
| **Pros** | Zero third-party dependencies (pure Python regex); fast; supports GFM tables, ATX headings, and local images (base64-encoded); deterministic — same file always produces the same tree |
| **Cons** | Only ATX headings (`#`) detected — setext-style headings (`===` underline) are not supported; remote image URLs are not fetched (stored as-is) |
| **vs. mistune / markdown-it-py** | Those libraries parse Markdown to a full AST — richer but overkill for RAG chunking; MarkdownParser extracts only what is needed for chunking and embedding |
| **vs. PdfParser / DocxParser** | Markdown structure is explicit in the source text, so tree construction is always reliable — no heuristics or style inspection required |

## Imports

```python
# Standard Library
import pathlib

# Third Party Library

# Private Library
from cleave.parsers.factory import ParserFactory
from cleave.schemas import ContentType, Document
```

## Fixture

```python
FIXTURES = pathlib.Path.cwd().parent.parent / "tests" / "fixtures"
MD_PATH = str(FIXTURES / "sample.md")

if not (FIXTURES / "sample.md").exists():
    from tests.fixtures.md import create_sample_md
    create_sample_md(FIXTURES / "sample.md")

print("MD:", MD_PATH)
```

| Mode | `Document` field | Use when |
|------|-----------------|----------|
| `flat` | `.pages` — one virtual `DocumentPage` | Simple text retrieval, fixed or sentence chunking |
| `tree` | `.root` — recursive `TreeNode` hierarchy | Semantic / heading-scoped chunking, multimodal pipelines |

## Flat mode

```python
flat_doc = ParserFactory.create(MD_PATH, mode="flat").parse()
assert isinstance(flat_doc, Document)
print(f"Source type : {flat_doc.source.type}")
print(f"Pages       : {len(flat_doc.pages)}")
print(f"Blocks      : {len(flat_doc.pages[0].blocks)}")
```

```python
for block in flat_doc.pages[0].blocks:
    preview = block.content[:60].replace("\n", "\\n")
    print(f"  [{block.position}] type={block.type.value:<6}  {preview!r}")
```

```python
tables = flat_doc.all_tables
print(f"Tables found: {len(tables)}")
print(tables[0].content)
```

## Tree mode

```python
tree_doc = ParserFactory.create(MD_PATH, mode="tree").parse()
assert tree_doc.root is not None
print(f"Root children : {len(tree_doc.root.children)}")
```

```python
def print_tree(node, indent=0):
    role = node.metadata.get("role", "")
    level = node.metadata.get("level", "")
    tag = f"{role}" + (f"[{level}]" if level else "")
    preview = node.content[:50].replace("\n", "\\n")
    print(" " * indent + f"{tag:<18}  {preview!r}")
    for child in node.children:
        print_tree(child, indent + 2)

print_tree(tree_doc.root)
```

```python
def find_headings(node):
    out = []
    if node.metadata.get("role") == "heading":
        out.append((node.metadata["level"], node.content))
    for child in node.children:
        out.extend(find_headings(child))
    return out

headings = find_headings(tree_doc.root)
print("Headings found:")
for lvl, text in headings:
    print(f"  H{lvl}: {text}")
```

```python
table_nodes = tree_doc.root.get_nodes_by_type(ContentType.table)
print(f"Table nodes : {len(table_nodes)}")
print(table_nodes[0].content)
```

## Bridge - tree to Markdown string

```python
md_out = tree_doc.to_markdown()
print(md_out)
```
